```python
from transformers import BertTokenizer

import matplotlib.pyplot as plt

# load BERT tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
```

```python
all_tokens = [token for token in tokenizer.get_vocab().keys() if "unused" not in token]

print(all_tokens)
```

```python
characters = ["1", "2", "3", "4", "5", "6", "7", "8", "9", 
              "a", "b", "c", "d", "e", "f", "g", "h", "i",
              "j", "k", "l", "m", "n", "o", "p", "q", "r",
              "s", "t", "u", "v", "w", "x", "y", "z"]

char_count = {}

for char in characters:
    char_count[char] = 0

print(char_count)
```

```python
for token in all_tokens:
    for char in token:
        if char in char_count.keys():
            char_count[char] += 1

print(char_count)


```
