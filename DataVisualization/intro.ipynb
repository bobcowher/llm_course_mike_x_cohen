```python
x = np.linspace(0, 3*np.pi, 1001)
y = np.sin(x)

plt.plot(x, y, 'k', label='A pretty sine wave')
plt.legend()
plt.show()
```

```python
fig, axs = plt.subplots(1, 3, figsize=(12, 4))
axs[0].plot(x, x**2, 'b')
```

```python
_, axs = plt.subplots(2, 2, figsize=(5,5))
axs[0,0].plot(np.random.randn(4,4))
axs[0,1].plot(np.random.randn(4,4))
axs[1,0].plot(np.random.randn(4,4))
axs[1,1].plot(np.random.randn(4,4))

plt.tight_layout()
plt.show()
```

```python
_, axs = plt.subplots(1, 3, figsize=(13.1, np.pi))

for exp, axis in enumerate(axs):
    axis.plot(x, x**exp)

plt.show()
```
