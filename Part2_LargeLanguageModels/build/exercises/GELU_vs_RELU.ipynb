```python
import torch
import torch.nn.functional as F
import numpy as np
from scipy.special import erf
import matplotlib.pyplot as plt
```

```python
def relu(x):
    y = np.maximum(np.zeros_like(x), x)
    return y

def gelu(x):
    y = x / np.sqrt(2)
    y = erf(y)
    y = 1 + y
    y = x * (0.5 * y) 
    return y


def gelu_approx(x):
    y = (0.5 * x) * (1 + np.tanh(np.sqrt(2/np.pi) * (x + (0.044715 * (np.pow(x, 3))))))
    return y
    
```

```python
print(relu(5))
print(gelu_approx(5))
print(gelu(5))
```

```python
x = np.linspace(-3, 3, 101)

y_relu = relu(x)
y_gelu = gelu(x)
y_gelu_approx = gelu_approx(x)
_
```

```python
plt.plot(x, y_relu, label="Relu")
plt.plot(x, y_gelu, label="Gelu")
plt.plot(x, y_gelu_approx, 'bo', markevery=5, label="Gelu Approx")

plt.legend()
plt.xlabel("X Values")
plt.ylabel("Non-Linear Function Output")
plt.title("Comparing Relu & Gelu")

plt.show()
```

```python
x = torch.from_numpy(x)

```

```python

y_relu_t = F.relu(x)
y_gelu_t = F.gelu(x)

plt.plot(x, y_relu_t, label="Relu")
plt.plot(x, y_gelu_t, label="Gelu")

plt.legend()
plt.xlabel("X Values")
plt.ylabel("Function Output")
plt.title("Comparing Relu & Gelu from Pytorch")

plt.show()
```

```python


y_relu_t_diff = torch.diff(y_relu_t)

print(len(y_relu_t))
print(len(y_relu_t_diff))

print(y_relu_t)
print(y_relu_t_diff)

# plt.plot(y_relu_t, torch.diff(y_relu_t))
```

```python
plt.plot(x[:-1], y_relu_t_diff)
plt.plot(x[:-1], torch.diff(y_gelu_t))
```
