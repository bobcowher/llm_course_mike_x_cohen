```python
import torch
import matplotlib.pyplot as plt
```

```python
data = torch.tensor([1.0, 5.0, 3.0, 2.0], dtype=torch.float32)

print(data)
```

```python
idx = torch.multinomial(data, 1)

print(f"Index: {idx[0]}")

value = data[idx]
print(f"Selected value: {value}")
```

```python
data = torch.linspace(0, 1, 20)

print(data)
```

```python
softmaxes = []

for i in range(8):
    data = torch.softmax(data, dim=0)
    print(f"Iteration {i} - Std {torch.std(data)}")
    softmaxes.append(data)

# print(softmaxes)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(15, 4))

print(softmaxes[0])

for i in range(8):
    ax[0].plot(torch.arange(0, len(softmaxes[i] + 1)), softmaxes[i], label=f"Iteration {i + 1}")
    ax[1].plot(i, torch.std(softmaxes[i]), label=f"Iteration {i + 1}", marker='o')

ax[0].title("Softmaxed data")
ax[1].title("Impact of softmaxes on variability")

ax[0].legend()
ax[1].legend()

```
