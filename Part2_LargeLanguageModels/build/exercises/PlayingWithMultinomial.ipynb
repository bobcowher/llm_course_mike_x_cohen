```python
import torch
import matplotlib.pyplot as plt
```

```python
data = torch.tensor([1.0, 5.0, 3.0, 2.0], dtype=torch.float32)

print(data)
```

```python
data = torch.linspace(0, 1, 20)

print(data)
```

```python
softmaxes = []

for i in range(8):
    data = torch.softmax(data, dim=0)
    print(f"Iteration {i} - Std {torch.std(data)}")
    softmaxes.append(data)

# print(softmaxes)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(15, 4))

print(softmaxes[0])

for i in range(8):
    ax[0].plot(torch.arange(0, len(softmaxes[i] + 1)), softmaxes[i], label=f"Iteration {i + 1}")
    ax[1].plot(i, torch.std(softmaxes[i]), label=f"Iteration {i + 1}", marker='o')

ax[0].set_title("Softmaxed data")
ax[1].set_title("Impact of softmaxes on variability")

ax[0].legend()
ax[1].legend()

```

```python
small_set_softmax = []
medium_set_softmax = []
large_set_softmax = []

temps = [0.5, 1, 3]

for idx,temp in enumerate(temps):

    small_set = torch.cat([torch.linspace(-0.4, 0.4, 100), torch.tensor([6])], dim=0)
    medium_set = torch.cat([torch.linspace(-1.0, 1.0, 100), torch.tensor([6])], dim=0)
    large_set = torch.cat([torch.linspace(-5.0, 5, 100), torch.tensor([6])], dim=0)

    small_set_scaled = small_set / temp
    medium_set_scaled = medium_set / temp
    large_set_scaled = large_set / temp

    small_set_softmax.append(torch.softmax(small_set_scaled, dim=0))
    medium_set_softmax.append(torch.softmax(medium_set_scaled, dim=0))
    large_set_softmax.append(torch.softmax(large_set_scaled, dim=0))


len(small_set_softmax)
```

```python
num_plots = len(small_set_softmax)

fig, ax = plt.subplots(1, num_plots, figsize=(15, 4))

for i, temp in enumerate(temps):
    ax[i].plot(small_set, small_set_softmax[i], label=f"-0.4 to 0.4")
    ax[i].plot(small_set, medium_set_softmax[i], label=f"-1.0 to 1.0")
    ax[i].plot(small_set, large_set_softmax[i], label=f"-5.0 to 5.0")
    ax[i].set_title(f"Softmaxed data - Temp {temp}")
    ax[i].legend()
```
