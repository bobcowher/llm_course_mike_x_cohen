```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
matrix_a = np.random.randint(0,100,(50,50))
matrix_b = np.random.randint(0,100,(50,50))

matrix_ab = matrix_a * matrix_b

print(matrix_ab.shape)
```

```python
print(f"Matrix A standard deviation: {np.std(matrix_a)}")
print(f"Matrix B standard deviation: {np.std(matrix_b)}")
print(f"Matrix AB standard deviation: {np.std(matrix_ab)}")
```

```python
print(f"Matrix A Dimensionality: {matrix_a.ndim}")
print(f"Matrix B Dimensionality: {matrix_b.ndim}")
print(f"Matrix AB Dimensionality: {matrix_ab.ndim}")
```

```python
plt.scatter(matrix_a[:, 0], matrix_a[:, 1], c="red")
plt.scatter(matrix_b[:, 0], matrix_b[:, 1], c="blue")
```

```python

plt.scatter(matrix_ab[:, 0], matrix_ab[:, 1], c="green")
```
